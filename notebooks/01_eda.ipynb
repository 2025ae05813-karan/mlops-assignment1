## 4. Key observations

- **Dataset size:** 303 patients, 13 features + binary target.
- **Class balance:** ~54% no disease / ~46% disease — reasonably balanced, so accuracy is
  a usable metric but we still report precision/recall/ROC-AUC.
- **Missing values:** only a handful of cells in `ca` and `thal`; handled by imputation
  inside the preprocessing pipeline (fit on training folds only to avoid leakage).
- **Correlations:** `cp`, `thalach`, `exang`, `oldpeak`, `ca`, and `thal` show the
  strongest relationships with the target — consistent with clinical expectations.
- **Modeling implication:** a mix of numeric and categorical features means we need
  scaling for numeric columns and one-hot encoding for categoricals, bundled into a single
  reproducible pipeline (next task).
